```python
#filter passengers older than 60

print('Passengers Older than 60:')
print(df[df["Age"]>60])
```

```python
#Passengers who survived

print("Passengers who survived")
print(df[df["Survived"]==1][['Name','Sex','Age']].head())
```

```python
#count survivors and non survivors

print("Survival counts:")
print(df['Survived'].value_counts())
```

```python
#calculte survival rate by gender
print("\nSurvival Rate by gender")
print(df.groupby("Sex")["Survived"].mean() * 100)
```

```python
#sort passengers by fare
print("Top 10 Highest Fares:")
print(
    df.sort_values("Fare",ascending=False)
    [["Name","Fare"]]
    .head(10)
)
```

```python
#count passenger by passenger class
print("Passengers by Class:")
print(df["Pclass"].value_counts().sort_index())
```

```python
#calculate age by gender
print("Average age by gender:")
print(df.groupby("Sex")["Age"].mean())
```

```python
#remove duplicates
df=df.drop_duplicates()
print("Duplicate rows removed")
```

```python
#Handle Missing Values
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col]=df[col].fillna(df[col].mean())
        else:
            df[col].fillna(df[col].mode()[0])

print("Missing Values handled")
```
